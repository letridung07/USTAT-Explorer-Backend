# understat.player(player) module

## _get_data()
```txt
player_data: dict
├── "player": dict
│   ├── basic player information
│   ├── usually player id/name info
│   └── useful as one-row metadata
│
├── "matches": list[dict]
│   ├── one dict = one match played by this player
│   ├── "goals": str
│   ├── "shots": str
│   ├── "xG": str
│   ├── "time": str
│   ├── "position": str
│   ├── "h_team": str
│   ├── "a_team": str
│   ├── "h_goals": str
│   ├── "a_goals": str
│   ├── "date": str
│   ├── "id": str              # match id
│   ├── "season": str
│   ├── "roster_id": str
│   ├── "xA": str
│   ├── "assists": str
│   ├── "key_passes": str
│   ├── "npg": str
│   ├── "npxG": str
│   ├── "xGChain": str
│   └── "xGBuildup": str
│
├── "shots": list[dict]
│   ├── one dict = one shot by this player
│   ├── "id": str
│   ├── "minute": str
│   ├── "result": str
│   ├── "X": str
│   ├── "Y": str
│   ├── "xG": str
│   ├── "player": str
│   ├── "h_a": "h" or "a"
│   ├── "player_id": str
│   ├── "situation": str
│   ├── "season": str
│   ├── "shotType": str
│   ├── "match_id": str
│   ├── "h_team": str
│   ├── "a_team": str
│   ├── "h_goals": str
│   ├── "a_goals": str
│   ├── "date": str
│   ├── "player_assisted": str | None
│   └── "lastAction": str
│
└── "groups": dict
    ├── "season": list[dict]
    │   ├── one dict = one season total
    │   ├── "position": str
    │   ├── "games": str
    │   ├── "goals": str
    │   ├── "shots": str
    │   ├── "time": str
    │   ├── "xG": str
    │   ├── "assists": str
    │   ├── "xA": str
    │   ├── "key_passes": str
    │   ├── "season": str
    │   ├── "team": str
    │   ├── "yellow": str
    │   ├── "red": str
    │   ├── "npg": str
    │   ├── "npxG": str
    │   ├── "xGChain": str
    │   └── "xGBuildup": str
    │
    ├── "position": dict
    │   └── grouped by season, then position
    │
    ├── "situation": dict
    │   └── grouped by season, then shot situation
    │       ├── "OpenPlay"
    │       ├── "FromCorner"
    │       ├── "SetPiece"
    │       ├── "DirectFreekick"
    │       └── "Penalty"
    │
    ├── "shotZones": dict
    │   └── grouped by season, then shot zone
    │       ├── "shotOboxTotal"
    │       ├── "shotPenaltyArea"
    │       └── "shotSixYardBox"
    │
    └── "shotTypes": dict
        └── grouped by season, then shot type
            ├── "RightFoot"
            ├── "LeftFoot"
            ├── "Head"
            └── "OtherBodyPart"
```
