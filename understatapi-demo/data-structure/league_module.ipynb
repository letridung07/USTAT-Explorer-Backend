# understat.league(league) module

## _get_data(season)
```txt
league_data: dict
├── "teams": dict
│   ├── "<team_id>": dict
│   │   ├── "id": str
│   │   ├── "title": str
│   │   └── "history": list[dict]
│   │       ├── one dict = one match played by that team
│   │       ├── "h_a": "h" or "a"
│   │       ├── "xG": float
│   │       ├── "xGA": float
│   │       ├── "npxG": float
│   │       ├── "npxGA": float
│   │       ├── "ppda": {"att": int, "def": int}
│   │       ├── "ppda_allowed": {"att": int, "def": int}
│   │       ├── "deep": int
│   │       ├── "deep_allowed": int
│   │       ├── "scored": int
│   │       ├── "missed": int
│   │       ├── "xpts": float
│   │       ├── "result": "w" / "d" / "l"
│   │       ├── "date": str
│   │       ├── "wins": int
│   │       ├── "draws": int
│   │       ├── "loses": int
│   │       ├── "pts": int
│   │       └── "npxGD": float
│
├── "players": list[dict]
│   ├── one dict = one player’s season total
│   ├── "id": str
│   ├── "player_name": str
│   ├── "games": str
│   ├── "time": str
│   ├── "goals": str
│   ├── "xG": str
│   ├── "assists": str
│   ├── "xA": str
│   ├── "shots": str
│   ├── "key_passes": str
│   ├── "yellow_cards": str
│   ├── "red_cards": str
│   ├── "position": str
│   ├── "team_title": str
│   ├── "npg": str
│   ├── "npxG": str
│   ├── "xGChain": str
│   └── "xGBuildup": str
│
└── "dates": list[dict]
    ├── one dict = one match / fixture
    ├── "id": str
    ├── "isResult": bool
    ├── "h": {"id": str, "title": str, "short_title": str}
    ├── "a": {"id": str, "title": str, "short_title": str}
    ├── "goals": {"h": str|null, "a": str|null}
    ├── "xG": {"h": str|null, "a": str|null}
    ├── "datetime": str
    └── "forecast": {"w": str, "d": str, "l": str}   # usually only for played matches
```

````python
from understatapi import UnderstatClient
LEAGUE = 'EPL'
SEASON = '2022'

with UnderstatClient() as understat:
    data = understat.league(LEAGUE)._get_data(SEASON)
    print(data['teams'].keys())
    print(data['teams']['71'].keys())
    print(data['teams']['71']['history'][0].keys())
    print(data['teams']['71']['history'][0]['date'])

    print(data['dates'][0]['datetime'])
    
````
