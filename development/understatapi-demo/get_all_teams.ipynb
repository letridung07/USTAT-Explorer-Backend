Fetch all teams from all seasons in a league

understat.league(league).get_team_data(season) data structure:
```txt
league_data: dict
├── "teams": dict
│   ├── "<team_id>": dict
│   │   ├── "id": str
│   │   ├── "title": str
│   │   └── "history": list[dict]
│   │       ├── ...
```

````python
from understatapi import UnderstatClient

LEAGUE = "EPL"
SEASON = ["2014", "2015", "2016", "2017", "2018", "2019", "2020", "2021", "2022", "2023", "2024", "2025"]
````

````python
all_teams = []
with UnderstatClient() as understat:
    for season in SEASON:
        all_teams_data = understat.league(LEAGUE).get_team_data(season)
        # All keys (team ids)
        print(all_teams_data.keys())
        
        # Extract team names from keys
        teams_1_season = []
        for key in all_teams_data.keys():
            teams_1_season.append(all_teams_data[key]["title"])
            
        print(f'season {season} {teams_1_season}')

        # Merge all teams from all seasons into a single list
        all_teams = list(set(all_teams + teams_1_season)) # Use set to remove duplicates

print(all_teams)

        
````
